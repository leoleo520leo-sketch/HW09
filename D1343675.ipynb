```python
# --- 測試一：原始資料 ---

# 1. 讀取活動名稱與原始資料
activity_names = ("桌遊體驗", "電影欣賞")
board_event, movie_event = activity_names  # tuple 拆包

board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 清理字串：移除半形空白、轉小寫，再依逗號拆成 list（保留重複）
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 使用 set 去重並進行集合運算
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set          # 交集：共同參加者
all_set = board_set | movie_set           # 聯集：全部參加者
only_board_set = board_set - movie_set    # 差集：只參加桌遊者
only_movie_set = movie_set - board_set    # 差集：只參加電影者

# 3. 排序名單並用切片取得通知名單前兩位
all_sorted = sorted(all_set)
notify_preview = all_sorted[:2]

# 4. 建立 summary 字典計算人數與應收費用
board_count = len(board_set)
movie_count = len(movie_set)
board_fee = unit_fees[board_event]
movie_fee = unit_fees[movie_event]

summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": board_count + movie_count,
    "桌遊應收": board_count * board_fee,
    "電影應收": movie_count * movie_fee,
    "總應收": (board_count * board_fee) + (movie_count * movie_fee)
}

# 5. 使用 f-string 輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{board_event}，每人 {board_fee:.2f} 元")
print(f"活動：{movie_event}，每人 {movie_fee:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(board_set)}")
print(f"電影去重名單：{sorted(movie_set)}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{sorted(both_set)}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{sorted(only_board_set)}")
print(f"只參加電影者：{sorted(only_movie_set)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
